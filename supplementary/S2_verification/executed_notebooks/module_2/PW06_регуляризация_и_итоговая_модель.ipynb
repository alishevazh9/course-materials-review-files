```python
import pandas as pd, joblib
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
df=pd.read_csv("../../data/fish_growth_dataset.csv");X=df[["age_weeks","mean_water_temp_C","daily_feed_ration_g","length_cm"]];y=df.weight_g
g=GridSearchCV(make_pipeline(StandardScaler(),Ridge()),{"ridge__alpha":[.1,1,10,100]},cv=5,scoring="neg_mean_absolute_error").fit(X,y)
print(g.best_params_, -g.best_score_); joblib.dump(g.best_estimator_,"growth_model.joblib")
```
