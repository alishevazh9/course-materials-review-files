## Оценивание

Используйте общую аналитическую рубрику из `assessments/rubric.md`.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
df=pd.read_csv("../../data/sturgeon_risk_classification.csv");X=df.drop(columns="risk");y=df.risk
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
m=LogisticRegression(max_iter=2000,class_weight="balanced").fit(Xtr,ytr)
print(classification_report(yte,m.predict(Xte),target_names=["normal","risk"]))
```
