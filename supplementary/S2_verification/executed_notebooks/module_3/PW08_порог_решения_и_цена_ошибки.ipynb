Используйте общую аналитическую рубрику из `assessments/rubric.md`.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_recall_fscore_support,confusion_matrix,average_precision_score,roc_auc_score
df=pd.read_csv("../../data/sturgeon_risk_classification.csv");X=df.drop(columns="risk");y=df.risk
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
m=LogisticRegression(max_iter=2000,class_weight="balanced").fit(Xtr,ytr); pr=m.predict_proba(Xte)[:,1]
print("tau precision recall F1 FP FN")
for tau in [.3,.5,.6,.7,.8]:
 p,r,f,_=precision_recall_fscore_support(yte,pr>=tau,average="binary",zero_division=0); tn,fp,fn,tp=confusion_matrix(yte,pr>=tau).ravel()
 print(tau,round(p,3),round(r,3),round(f,3),fp,fn)
print("Average precision",round(average_precision_score(yte,pr),3),"ROC AUC",round(roc_auc_score(yte,pr),3))
```
