```python
import pandas as pd, numpy as np
from sklearn.preprocessing import MinMaxScaler
df=pd.read_csv("../../data/hydrochem_timeseries.csv",parse_dates=["timestamp"]).set_index("timestamp")
a=df[["temperature_C","dissolved_oxygen_mgL","pH","ammonia_mgL"]].values; cut=int(len(a)*.8)
sc=MinMaxScaler().fit(a[:cut]); z=sc.transform(a)
W,H=24,6; X=[];y=[]
for i in range(len(z)-W-H+1): X.append(z[i:i+W]);y.append(z[i+W+H-1,1])
X,y=np.array(X),np.array(y); split=max(0,cut-W-H+1)
print("Train/test windows",X[:split].shape,X[split:].shape)
# Обучение TensorFlow выполняется студентом; сравнение проводится на тех же временных точках, что и ARIMA.
```
